## 1. Business Scenario
You have been hired as a **Junior Data Scientist** by a retail company. The Marketing Department has
collected two years’ worth of data on their customers but lacks the analytical framework to use it
effectively.

Your goal is to perform a complete end-to-end analysis - from assessing data integrity to identifying
high-value customer segments - to drive a data-driven communication strategy.

## 3. Exam requirements
### Task A: Audit & Quality
- Perform a Data Audit: identify missing values, duplicates and data type inconsistencies
- Identify and handle Outliers
- Document your cleaning process: Why did you choose to drop, impute, or transform specific data points?
### Task B: Exploratory Data Analysis
- Analyze the correlation between socio-demographic features and total spending
- Visualize the distribution of purchases across different channels
- Communication Insight: Identify one "hidden" trend in the data that could be used for a targeted promotional campaign
### Task C: RFM Calculation & Heuristic Segmentation
- Engineer the Recency, Frequency, and Monetary features
- Assign scores to customers based on quintiles
- Define at least 4 segments based on these scores
### Task D: Unsupervised Learning (Cluster Analysis)
- Apply the K-Means algorithm to segment customers
- Justify the optimal number of clusters using the Elbow Method or Silhouette Score
- Compare the Clusters with your RFM segments. Do they overlap? What additional information do the clusters capture that RFM might have missed?
### Task E: Communication Strategy
- Based on your findings, suggest a specific Communication Channel and Message Tone for your identified segments
